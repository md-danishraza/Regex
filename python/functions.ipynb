a) re.search()

Finds first occurrence anywhere in string.

```python
import re

text = "I love Python"

result = re.search("Python", text)

print(result.group())
```

b) re.match()

Checks pattern only at the beginning.

```python
import re

text = "Python Programming"

print(re.match("Python", text))
```

c) re.fullmatch()

Entire string must match.

```python
import re

print(re.fullmatch(r"\d+", "12345"))
```

d) re.findall()

```python
import re

text = "cat bat rat"

print(re.findall("at", text))
```

re.finditer()
Returns iterator of match objects.

```python
import re

text = "123 456 789"

for match in re.finditer(r"\d+", text):
    print(match.group())
```

f) re.sub()
Replace text.

```python
import re

text = "I like Java"

print(re.sub("Java", "Python", text))
```

g) re.split()

Split using regex.

```python
import re

text = "apple,orange;banana"

print(re.split(r"[,;]", text))
```
